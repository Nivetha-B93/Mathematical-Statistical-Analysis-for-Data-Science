Dataset:
Linear Algebra Dataset:

```python
import numpy as np
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
[2, 1, 3],
[1, 0, 2],
[4, 1, 1]
])
matrix_b = np.array([
[1, 2, 0],
[3, 1, 4],
[2, 5, 1]
])
```

Using NumPy:
● Perform vector addition, subtraction, and dot product using vector_a and vector_b

```python
vector_addition = vector_a + vector_b
print("Vector Addition:", vector_addition)
```

```python
vector_subtraction = vector_a - vector_b
print("Vector Subtraction:", vector_subtraction)
```

```python
dot_product = np.dot(vector_a, vector_b)
print("Dot Product:", dot_product)
```

● Perform matrix addition and matrix multiplication using matrix_a and matrix_b

```python
matrix_addition = matrix_a + matrix_b
print("Matrix Addition:\n", matrix_addition)
```

```python
matrix_multiplication = np.dot(matrix_a, matrix_b)
print("Matrix Multiplication:\n", matrix_multiplication)
```

● Calculate determinant of matrix_a

```python
determinant = np.linalg.det(matrix_a)
print("Determinant of Matrix A:", determinant)
```

If determinant of matrix_a is not equal to zero, then find inverse of matrix_a

```python
determinant = np.linalg.det(matrix_a)

if determinant != 0:
    inverse_matrix = np.linalg.inv(matrix_a)
    print("Inverse of Matrix A:\n", inverse_matrix)
else:
    print("Matrix A does not have an inverse.")
```

● Compute eigenvalues and eigenvectors of matrix_a

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

Apply Bayes’ theorem to calculate updated probability of a customer purchase
○ Formula: P(A|B) = [P(B|A) * P(A)] / P(B)

```python
# Given probabilities
P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80

# Bayes' theorem
P_A_given_B = (P_B_given_A * P_A) / P_B

print("P(A|B) =", P_A_given_B)
print("Probability =", P_A_given_B * 100, "%")
```

Task 3 – Descriptive Statistics
Statistical Dataset

```python
sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]
```

Compute mean, median, and mode of sales_data

```python
import numpy as np
import pandas as pd
import statistics

sales_data = [
    120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
    190, 200, 210, 195, 185, 175, 165, 155, 145, 135
]

# Mean
mean = np.mean(sales_data)

# Median
median = np.median(sales_data)

# Mode
mode = statistics.mode(sales_data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
mode = statistics.multimode(sales_data)

print("Mode:", mode)
```

Calculate range, variance, and standard deviation

```python
# Range
data_range = np.max(sales_data) - np.min(sales_data)

# Variance
variance = np.var(sales_data)

# Standard Deviation
standard_deviation = np.std(sales_data)

print("Range:", data_range)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

Measure skewness and kurtosis

```python
from scipy.stats import skew, kurtosis

skewness = skew(sales_data)
kurtosis_value = kurtosis(sales_data)

print("Skewness:", skewness)
print("Kurtosis:", kurtosis_value)
```

Interpret whether the dataset is symmetric, positively skewed, or negatively skewed

For this sales_data, the skewness is approximately 0.

Interpretation
Skewness ≈ 0

Therefore, the dataset is approximately symmetric.


Skewness > 0 → Positively (right) skewed
Skewness < 0 → Negatively (left) skewed
Skewness ≈ 0 → Symmetric

This data has:

Mean = 165
Median = 165
Skewness ≈ 0

Since the mean and median are equal and skewness is approximately zero, the sales distribution is symmetric.

Conclusion: ✅ The dataset is approximately symmetric, not positively or negatively skewed.
